```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
Data=pd.read_csv('/content/drive/MyDrive/Copy of Load_Extension - Sheet1.csv')
print(Data)
```

```python
x=Data[["Load (N)"]]
y=Data[["Extension (mm)"]]
```

```python
plt.scatter(x,y)
plt.xlabel("Load")
plt.ylabel("Extension")
plt.title("Load Vs Extension")
plt.show()
```

```python
model=LinearRegression()
model.fit(x,y)
```

```python
print("Coefficient:",model.coef_)
print("intercept:",model.intercept_)
```

```python
load=float(input("Enter Load"))
Extension= model.predict([[load]])
print("Predicted Extension :",Extension)
```
